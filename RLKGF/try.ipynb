```python
import json
import csv
import re
import os

def generate_csv_for_neo4j():
    """
    Loads data from Dataset.txt and creates Neo4j-import-compatible CSV files
    for nodes (problems, slots) and their relationships.
    """
    input_filename = 'Dataset.txt'
    
    print("Starting CSV generation for Neo4j...")

    if not os.path.exists(input_filename):
        print(f"Error: '{input_filename}' not found.")
        return

    # --- Load and Clean the Dataset ---
    try:
        with open(input_filename, 'r', encoding='utf-8') as f:
            content = f.read()
            match = re.search(r'\[\s*\{.*\}\s*\]', content, re.DOTALL)
            if match:
                data = json.loads(match.group(0))
                print(f"Successfully loaded {len(data)} records.")
            else:
                print("Error: Could not find a valid JSON list in the file.")
                return
    except Exception as e:
        print(f"An error occurred while parsing the file: {e}")
        return

    # --- Collect all unique nodes and relationships ---
    all_problems = set()
    all_slots = set()
    relationships = set()

    for record in data:
        problem = record.get('problem_tag')
        if not problem:
            continue
        
        all_problems.add(problem)
        
        slots = {**record.get('explicit_inform_slots', {}), **record.get('implicit_inform_slots', {})}
        for slot, is_present in slots.items():
            if is_present:
                all_slots.add(slot)
                relationships.add((problem, slot))

    # --- Create mapping from name to a unique ID for Neo4j ---
    problem_to_id = {name: f"p_{i}" for i, name in enumerate(all_problems)}
    slot_to_id = {name: f"s_{i}" for i, name in enumerate(all_slots)}
    
    # --- Write Node Files ---
    # Problems
    with open('problems_header.csv', 'w', newline='', encoding='utf-8') as f:
        f.write('problemId:ID,name,:LABEL\n')
    with open('problems.csv', 'w', newline='', encoding='utf-8') as f:
        writer = csv.writer(f)
        for name, uid in problem_to_id.items():
            writer.writerow([uid, name, 'Problem'])
    print(f"Generated problems.csv with {len(all_problems)} nodes.")

    # Slots
    with open('slots_header.csv', 'w', newline='', encoding='utf-8') as f:
        f.write('slotId:ID,name,:LABEL\n')
    with open('slots.csv', 'w', newline='', encoding='utf-8') as f:
        writer = csv.writer(f)
        for name, uid in slot_to_id.items():
            writer.writerow([uid, name, 'Slot'])
    print(f"Generated slots.csv with {len(all_slots)} nodes.")

    # --- Write Relationship File ---
    with open('relationships_header.csv', 'w', newline='', encoding='utf-8') as f:
        f.write(':START_ID,:END_ID\n')
    with open('relationships.csv', 'w', newline='', encoding='utf-8') as f:
        writer = csv.writer(f)
        for problem_name, slot_name in relationships:
            start_id = problem_to_id[problem_name]
            end_id = slot_to_id[slot_name]
            writer.writerow([start_id, end_id])
    print(f"Generated relationships.csv with {len(relationships)} relationships.")
    
    print("\nCSV generation complete! Files are ready for Neo4j import.")

if __name__ == '__main__':
    generate_csv_for_neo4j()
```

```python
import json
import re

def create_entity_files(input_path="Dataset.txt"):
    """
    Parses the dataset file to extract unique problem tags and symptoms,
    then writes them to their respective output files.

    Args:
        input_path (str): The path to the dataset text file.
    """
    try:
        with open(input_path, 'r', encoding='utf-8') as f:
            content = f.read()
    except FileNotFoundError:
        print(f"Error: The file '{input_path}' was not found.")
        return

    # Correctly remove lines using the valid regex
    cleaned_content = re.sub(r'\\', '', content)

    try:
        # Load the cleaned content as a JSON object (a list of dictionaries)
        data = json.loads(cleaned_content)
    except json.JSONDecodeError as e:
        print(f"Error: Could not parse the file as JSON. {e}")
        return

    problem_tags = set()
    symptoms = set()

    # Iterate through each record in the dataset
    for record in data:
        if "problem_tag" in record:
            problem_tags.add(record["problem_tag"])
        
        if "explicit_inform_slots" in record and isinstance(record["explicit_inform_slots"], dict):
            for symptom in record["explicit_inform_slots"].keys():
                symptoms.add(symptom)

    # Write the unique problem tags to omega_problems.txt
    with open("omega_problems.txt", 'w', encoding='utf-8') as f:
        # Sorting ensures a consistent order
        for tag in sorted(list(problem_tags)):
            f.write(tag + '\n')
            
    print(f"✅ Successfully created 'omega_problems.txt' with {len(problem_tags)} unique problem tags.")

    # Write the unique symptoms to omega_symptoms.txt
    with open("omega_symptoms.txt", 'w', encoding='utf-8') as f:
        # Sorting ensures a consistent order
        for symptom in sorted(list(symptoms)):
            f.write(symptom + '\n')
            
    print(f"✅ Successfully created 'omega_symptoms.txt' with {len(symptoms)} unique symptoms.")

# Run the function, assuming 'Dataset.txt' is in the same directory
create_entity_files("Dataset.txt")
```

```python
import json
import pickle
import random

def split_dataset(input_path="Dataset.txt", output_path="omega_goals.p", train_ratio=0.8):
    # Read the dataset file
    try:
        with open(input_path, 'r', encoding='utf-8') as f:
            data = json.load(f)  # assuming Dataset.txt is valid JSON
    except FileNotFoundError:
        print(f"Error: File '{input_path}' not found.")
        return
    except json.JSONDecodeError as e:
        print(f"Error: Failed to decode JSON. {e}")
        return

    # Shuffle the data for random splitting
    random.shuffle(data)

    # Calculate split index
    split_idx = int(len(data) * train_ratio)

    # Split the data
    train_data = data[:split_idx]
    test_data = data[split_idx:]

    # Create the final dictionary
    final_dict = {
        'train': train_data,
        'test': test_data
    }

    # Save to pickle file
    with open(output_path, 'wb') as f:
        pickle.dump(final_dict, f)

    print(f"✅ Successfully saved train/test split to '{output_path}'.")
    print(f"Train size: {len(train_data)}, Test size: {len(test_data)}")

# Run the function
split_dataset()
```

```python
import os
import torch
import torch.optim as optim
import pickle
import random
import numpy as np
import ast

# Assuming GCN.py and KG.py are in the same directory or accessible in the python path
from utils.utils import *
from utils import dialog_config
from Qwen25_3b_Test.DisPre_RWR_And_GCN.GCN import GCNReward
from Data.DXY.KG import KGADJ
from utils import dialog_config # Assuming this contains constants like TRUE/FALSE

# --- Helper Functions (to make the script self-contained) ---

def load_pickle(file):
    """Loads a pickle file."""
    with open(file, 'rb') as f:
        return pickle.load(f)

def text_to_list(file):
    """Reads a text file into a list of lines."""
    with open(file, 'r', encoding='utf-8') as f:
        return [line.strip() for line in f.readlines()]

def set_seed(seed):
    """Sets random seeds for reproducibility."""
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    if torch.cuda.is_available():
        torch.cuda.manual_seed(seed)
        torch.cuda.manual_seed_all(seed)
    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark = False

# --- Main Training Function ---

def train_gcn_reward_model():
    """
    Main function to load data, train the GCNReward model, and save the best version.
    """
    # --- 1. Configuration and Setup ---
    set_seed(42)
    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    print(f"Using device: {device}")

    # Hyperparameters
    epochs = 50
    learning_rate = 0.001
    batch_size = 16
    embed_size = 128 # Can be tuned

    # Define file paths for your new Omega dataset
    # Assumes these files are in a subdirectory named 'omega_data'
    data_path = './' # Change if your data is in a different folder
    problems_file = os.path.join(data_path, 'omega_problems.txt')
    observations_file = os.path.join(data_path, 'omega_symptoms.txt')
    goal_file = os.path.join(data_path, 'omega_goals.p')
    problem_obs_dict_file = os.path.join(data_path, 'prob_slot_num_dict_maintenance_weighted.txt')
    save_path = './omega_gcn_reward.pth.tar'

    # --- 2. Load Data and Build Knowledge Graph ---
    problems = text_to_list(problems_file)
    observations = text_to_list(observations_file)
    goals = load_pickle(goal_file)

    num_problems = len(problems)
    num_observations = len(observations)
    num_total_nodes = num_problems + num_observations

    # Create the slot_set dictionary mapping entities to indices
    slot_set = {**{prob: i for i, prob in enumerate(problems)}, 
                **{obs: i + num_problems for i, obs in enumerate(observations)}}

    # The KGADJ class needs the problem-symptom dictionary to build the adjacency matrix.
    # We rename the original files it expects internally for it to work.
    # This is a workaround to avoid modifying KG.py
    # if not os.path.exists('diseases_dxy.txt'):
    #     os.rename(problems_file, 'diseases_dxy.txt')
    # if not os.path.exists('symptoms_dxy.txt'):
    #     os.rename(observations_file, 'symptoms_dxy.txt')
    # if not os.path.exists('dise_sym_num_dict_dxy.txt'):
    #     with open(os.path.join(data_path, 'dise_sym_num_dict_dxy.txt'), 'r', encoding='utf-8') as f:
    #         content = f.read()
    #         self.dis_sys_pre = ast.literal_eval(content)


    kg = KGADJ(device, num_problems, num_observations, data_path)
    kg.initialize_adj() # This creates the adjacency matrix
    
    # Restore original filenames
    # os.rename('diseases_dxy.txt', problems_file)
    # os.rename('symptoms_dxy.txt', observations_file)
    # os.remove('dise_sym_num_dict_dxy.txt')


    # Convert the adjacency matrix to edge_index format for PyTorch Geometric
    adj_matrix = kg.kg_matrix
    edge_index = adj_matrix.nonzero().t().contiguous().to(device)

    # --- 3. Initialize Model and Optimizer ---
    model = GCNReward(
        device=device,
        kg_node=num_total_nodes,
        dis_num=num_problems,
        kggraph=kg,
        slot_set=slot_set,
        embed_size=embed_size
    ).to(device)

    optimizer = optim.Adam(model.parameters(), lr=learning_rate)
    
    print(f"Model initialized with {num_problems} problems and {num_observations} observations.")
    
    # --- 4. Training and Validation Loop ---
    best_accuracy = 0.0
    
    train_data = goals['train']
    test_data = goals['test']

    for epoch in range(epochs):
        model.train()
        total_loss = 0
        total_correct = 0
        total_samples = 0
        
        # Shuffle training data each epoch
        random.shuffle(train_data)
        
        # --- Training Phase ---
        for i in range(0, len(train_data), batch_size):
            batch = train_data[i:i+batch_size]
            if not batch:
                continue

            labels = []
            masks = []

            for goal_item in batch:
                print("Example train_data item:", train_data[0])
                print("Type:", type(train_data[0]))

                goal = goal_item['problem_tag']
                
                # Create label: the index of the correct problem
                problem_tag = goal_item['problem_tag']
                if problem_tag in problems:
                    labels.append(problems.index(problem_tag))
                else:
                    continue # Skip if problem tag is not found
                
                # Create mask: a vector indicating present observations
                mask = torch.zeros(num_total_nodes)
                # confirmed_obs = [k for k, v in goal['current_slots']['inform_slots'].items() if v == dialog_config.TRUE]
                confirmed_obs = []

                # Combine explicit and implicit slots with TRUE values
                for slot_dict in [goal_item.get('explicit_inform_slots', {}), goal_item.get('implicit_inform_slots', {})]:
                    confirmed_obs.extend([k for k, v in slot_dict.items() if v == dialog_config.TRUE])

                for obs in confirmed_obs:
                    if obs in slot_set:
                        mask[slot_set[obs]] = 1.0
                masks.append(mask)

            if not labels:
                continue

            labels_tensor = torch.LongTensor(labels).to(device)
            masks_tensor = torch.stack(masks).to(device)

            optimizer.zero_grad()
            #print("disease_ shape:", disease_.shape)   # Should be [batch_size, num_classes] e.g. [16, 345]
            print("labels_tensor shape:", labels_tensor.shape)  # Should be [batch_size] e.g. [16]
            print("labels_tensor:", labels_tensor)

            # Use the forward_before method for direct classification training
            correct, loss = model.forward_before(labels_tensor, edge_index, masks_tensor)
            
            loss.backward()
            optimizer.step()

            total_loss += loss.item() * len(batch)
            total_correct += correct
            total_samples += len(batch)
            
        avg_train_loss = total_loss / total_samples
        train_accuracy = total_correct / total_samples

        # --- Validation Phase ---
        model.eval()
        total_val_correct = 0
        total_val_samples = 0
        with torch.no_grad():
            for i in range(0, len(test_data), batch_size):
                batch = test_data[i:i+batch_size]
                if not batch:
                    continue

                labels = []
                masks = []
                for goal_item in batch:
                    #goal = goal_item['problem_tag']
                    problem_tag = goal_item['problem_tag']
                    if problem_tag in problems:
                        labels.append(problems.index(problem_tag))
                    else:
                        continue
                    
                    mask = torch.zeros(num_total_nodes)
                    # confirmed_obs = [k for k, v in goal['current_slots']['inform_slots'].items() if v == dialog_config.TRUE]
                    confirmed_obs = []

                    # Combine explicit and implicit slots with TRUE values
                    for slot_dict in [goal_item.get('explicit_inform_slots', {}), goal_item.get('implicit_inform_slots', {})]:
                        confirmed_obs.extend([k for k, v in slot_dict.items() if v == dialog_config.TRUE])

                    for obs in confirmed_obs:
                        if obs in slot_set:
                            mask[slot_set[obs]] = 1.0
                    masks.append(mask)

                if not labels:
                    continue
                    
                labels_tensor = torch.LongTensor(labels).to(device)
                masks_tensor = torch.stack(masks).to(device)
                
                correct, _ = model.forward_before(labels_tensor, edge_index, masks_tensor)
                total_val_correct += correct
                total_val_samples += len(batch)
        
        val_accuracy = total_val_correct / total_val_samples
        
        print(f"Epoch {epoch+1}/{epochs} | Train Loss: {avg_train_loss:.4f} | Train Acc: {train_accuracy:.4f} | Val Acc: {val_accuracy:.4f}")

        # --- 5. Save the Best Model ---
        if val_accuracy > best_accuracy:
            best_accuracy = val_accuracy
            torch.save({
                'state_dict': model.state_dict(),
                'best_accuracy': best_accuracy,
                'epoch': epoch + 1
            }, save_path)
            print(f"  -> New best model saved to {save_path} with accuracy: {best_accuracy:.4f}")

if __name__ == "__main__":
    train_gcn_reward_model()
```

```python

# DataLoader for batching
batch_size = 16
dataloader = DataLoader(dataset, batch_size=batch_size, shuffle=True)

# Initialize model
model = GCNReward(device=device,
                  kg_node=kg_adj.dis_num + kg_adj.sym_num,
                  dis_num=kg_adj.dis_num,
                  kggraph=kg_adj,
                  slot_set={**{d: i for i, d in enumerate(kg_adj.disease_pre)},
                            **{s: i + kg_adj.dis_num for i, s in enumerate(kg_adj.sys_pre)}},
                  embed_size=100).to(device)

optimizer = Adam(model.parameters(), lr=0.001)

# Helper function: convert dense adjacency matrix batch to edge_index list and edge_weight list
def batch_dense_to_sparse(adj_batch):
    # adj_batch: [B, N, N]
    edge_indices = []
    edge_weights = []
    for i in range(adj_batch.size(0)):
        edge_index, edge_weight = dense_to_sparse(adj_batch[i])
        edge_indices.append(edge_index)
        edge_weights.append(edge_weight)
    return edge_indices, edge_weights

# Modify your model's forward method to accept edge_index and edge_weight
def forward_with_edge_index(self, label, edge_index, mask, edge_weight=None):
    # x shape: [batch_size, num_nodes] = node indices, embedding lookup done inside model
    batch_size = mask.size(0)
    x = torch.tensor([[i for i in range(self.kg_node)] for _ in range(batch_size)]).to(self.device)
    x_embed = self.sym_representation(x)

    x_out = []
    for b in range(batch_size):
        h = F.relu(self.gc1(x_embed[b], edge_index[b], edge_weight[b] if edge_weight is not None else None))
        h = self.gc2(h, edge_index[b], edge_weight[b] if edge_weight is not None else None)
        x_out.append(h)

    x = torch.stack(x_out, dim=0)  # [batch_size, num_nodes, embed_size]

    x = x[:, :self.dis_num]  # only diseases, shape: [batch_size, dis_num, embed_size]

    mask_disease = mask[:, :self.dis_num]  # shape: [batch_size, dis_num]

    x = x * mask_disease.unsqueeze(-1)  # broadcast mask to [batch_size, dis_num, embed_size]

    # Now sum over embedding dimension for final prediction scores
    x = x.sum(dim=2)  # shape: [batch_size, dis_num]

    prob = torch.softmax(x, dim=1)
    loss = torch.nn.CrossEntropyLoss()(x, label)

    correct = (torch.argmax(prob, dim=1) == label).sum().item()


    return correct, loss


# Attach the new forward method to your model class
import types
model.forward = types.MethodType(forward_with_edge_index, model)

# Training loop
epochs = 10
model.train()

for epoch in range(epochs):
    total_loss = 0
    total_correct = 0
    total_samples = 0

    for labels, masks in dataloader:
        labels = labels.to(device)
        masks = masks.to(device)

        model.kgraph.initialize_adj()
        kg_matrix = model.kgraph.kg_matrix.unsqueeze(0).repeat(labels.size(0), 1, 1)

        edge_indices, edge_weights = batch_dense_to_sparse(kg_matrix)

        optimizer.zero_grad()

        batch_correct = 0
        batch_loss = 0
        for i in range(labels.size(0)):
            label_i = labels[i].unsqueeze(0)
            mask_i = masks[i].unsqueeze(0)
            edge_index_i = edge_indices[i].to(device)
            edge_weight_i = edge_weights[i].to(device)

            correct, loss = model.forward(label_i, [edge_index_i], mask_i, edge_weight=[edge_weight_i])
            loss.backward()
            optimizer.step()

            batch_correct += correct
            batch_loss += loss.item()

        total_loss += batch_loss
        total_correct += batch_correct
        total_samples += labels.size(0)

    print(f"Epoch {epoch+1}/{epochs}, Loss: {total_loss/total_samples:.4f}, Accuracy: {total_correct/total_samples:.4f}")
```
